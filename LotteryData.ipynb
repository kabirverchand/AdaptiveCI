```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pyreadr

import confidence_intervals
```

```python
result = pyreadr.read_r('data/lottery.RData')
# result
lottery_data = result['d']
```

```python
winner_data = lottery_data[lottery_data['winner'] == 1]
control_data = lottery_data[lottery_data['winner'] == 0]
```

```python
average_earnings_winner = []
indicator_pos_earnings_winner = []
for index, row in winner_data.iterrows():
  avg = 0
  avg += row["yearn.1"]
  avg += row["yearn.2"]
  avg += row["yearn.3"]
  avg += row["yearn.4"]
  avg += row["yearn.5"]
  avg += row["yearn.6"]
  avg += row["yearn.7"]
  avg = avg/7
  average_earnings_winner.append(avg)

  if row["yearn.1"] > 0 and row["yearn.2"] > 0 and row["yearn.3"] > 0 and row["yearn.4"] > 0 and row["yearn.5"] > 0 and row["yearn.6"] > 0 and row["yearn.7"] > 0:
    indicator_pos_earnings_winner.append(1)
  else:
    indicator_pos_earnings_winner.append(0)

average_earnings_control = []
indicator_pos_earnings_control = []
for index, row in control_data.iterrows():
  avg = 0
  avg += row["yearn.1"]
  avg += row["yearn.2"]
  avg += row["yearn.3"]
  avg += row["yearn.4"]
  avg += row["yearn.5"]
  avg += row["yearn.6"]
  avg += row["yearn.7"]
  avg = avg/7
  average_earnings_control.append(avg)

  if row["yearn.1"] > 0 and row["yearn.2"] > 0 and row["yearn.3"] > 0 and row["yearn.4"] > 0 and row["yearn.5"] > 0 and row["yearn.6"] > 0 and row["yearn.7"] > 0:
    indicator_pos_earnings_control.append(1)
  else:
    indicator_pos_earnings_control.append(0)
```

```python
### Make in missing format
average_earnings_winner_missing = np.append(average_earnings_winner, np.ones(len(average_earnings_control)) * np.nan)
average_earnings_control_missing = np.append(average_earnings_control, np.ones(len(average_earnings_winner)) * np.nan)

indicator_earnings_winner_missing = np.append(indicator_pos_earnings_winner, np.ones(len(average_earnings_control)) * np.nan)
indicator_earnings_control_missing = np.append(indicator_pos_earnings_control, np.ones(len(average_earnings_winner)) * np.nan)
```
